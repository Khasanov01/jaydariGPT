```python
text="Hello world this is a simple example to show how tokenization works in NLP and a sentence"
print(text)      # 17 tokens
```

```python
tokens=text.split()
print(tokens)
```

```python
vocab={}
current_id=0

for token in tokens:
  if token not in vocab:
    vocab[token]=current_id
    current_id+=1
print(vocab)
```

```python
encoded=[vocab[token] for token in tokens]
print(encoded)
```

```python
reverse_vocab={id:token for token,  id in vocab.items()}
decoded="".join([reverse_vocab[id] for id in encoded])
print(decoded)
```

```python
# Embedding vector => dense numerical array | Vector Database
# 0=[0.44, -0.21, 0.77, ...]
# 1=[0.56, 0.12, -0.34, ...]
```

```python
import torch
import torch.nn as nn

embedding_dim=3
vocab_size=len(vocab)

embedding_layer=nn.Embedding(vocab_size, embedding_dim)

input_ids=torch.tensor([encoded])
embeddings=embedding_layer(input_ids)

print(embeddings)
print(embeddings.shape)
```

```python
from transformers import AutoTokenizer

model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer=AutoTokenizer.from_pretrained(model_id)
```

```python
print("vocab_size:", tokenizer.vocab_size)
print("special tokens:", tokenizer.special_tokens_map)
```

```python
text="Hello how are you doing?"
print(text)
```

```python
tokens=tokenizer.tokenize(text)
print("tokens:", tokens)
```

```python
ids=tokenizer.convert_tokens_to_ids(tokens)
print("token IDs:", ids)
```

```python
encoded=tokenizer(text)
print(encoded)
```

```python
print("input IDs:", encoded["input_ids"])
print("attention mask:", encoded["attention_mask"])
```

```python
# "Hi" #=> [1, 345, 2, 2, 2] // length 5
# "How are you?" #=> [1, 45, 657, 234, 23453] // length 5
# 'attention_mask': [1, 1, 0, 0, 0]
```

```python
decoded_text=tokenizer.decode(encoded["input_ids"])
print("decoded text:", decoded_text)
```
